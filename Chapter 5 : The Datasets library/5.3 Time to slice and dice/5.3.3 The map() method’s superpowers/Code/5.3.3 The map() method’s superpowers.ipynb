```python
import pandas as pd
from datasets import Dataset
from transformers import AutoTokenizer

# 1. Load dataset
train_df = pd.read_csv("train_5000.tsv", delimiter="\t")

# 2. Convert pandas dataframe to HuggingFace dataset
dataset = Dataset.from_pandas(train_df)

# 3. Load tokenizer
tokenizer = AutoTokenizer.from_pretrained("bert-base-cased")

# 4. Tokenization function
def tokenize_function(examples):
    return tokenizer(examples["review"], truncation=True)

# 5. Apply tokenization
tokenized_dataset = dataset.map(tokenize_function, batched=True)

print(tokenized_dataset)
```

```python
# 7. Example of multiprocessing (optional)
tokenized_dataset = dataset.map(
    tokenize_function,
    batched=True,
    num_proc=8
)


# 8. Tokenization with long text splitting
def tokenize_and_split(examples):
    return tokenizer(
        examples["review"],
        truncation=True,
        max_length=128,
        return_overflowing_tokens=True
    )


# 9. Apply tokenization with splitting
tokenized_dataset_split = dataset.map(
    tokenize_and_split,
    batched=True,
    remove_columns=dataset.column_names
)


# 10. Print dataset information
print(tokenized_dataset)
print(tokenized_dataset_split)
```
